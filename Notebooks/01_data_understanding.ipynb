## 1. Import and Load Data

```python
import pandas as pd
import confidential_file as paths


customers = paths.customers
products = paths.products
sales = paths.sales
```

## 2. Basic Overview
Understand:
- Number of rows and columns
- First few records
- Column names
- Data types

```python
tables = {
    'customers': customers,
    'products': products,
    'sales': sales
}
```

```python
class basic_overview():
    def rows_columns_count(self):
        for name, df in tables.items():
            print(f"In {name} table:\nNumber of rows: {df.shape[0]}\nNumber of columns: {df.shape[1]}\n")

    def first_few_rows(self):
        for name, df in tables.items():
            print(f"First few record in {name} table:\n{df.head()}\n")

    def columns_name(self):
        for name, df in tables.items():
            print(f"In {name} table:\nColumns name: {df.columns}\n")

    def data_types(self):
        for name, df in tables.items():
            print(f"In {name} table:\nData types:\n{df.dtypes}\n")


obj = basic_overview()
```

```python
obj.rows_columns_count()
```

```python
obj.first_few_rows()
```

```python
obj.columns_name()
```

```python
obj.data_types()
```

## 3. Data Summary
Check:
- Numerical columns
- Minimum and maximum values
- Mean and median
- Unique values

```python
class data_summary():
    def show_numerical_col(self):
        for name, df in tables.items():
            print(f"In {name} table:\nNumerical columns: {df.select_dtypes(include=['number']).columns}\n")

    def minimum_maximum_values(self):
        if 'customers' in tables:
            print(f"In customers table")
            print(f"minimum age is {tables['customers']['Age'].min()}\nmaximum age is {tables['customers']['Age'].max()}")
            print(f"minimum order is {tables['customers']['Total_Orders'].min()}\nmaximum order is {tables['customers']['Total_Orders'].max()}")
            print(f"minimum spent is {tables['customers']['Total_Spent'].min()}\nmaximum spent is {round(tables['customers']['Total_Spent'].max(), 1)}")
            print() 
            print()

            
        if 'products' in tables:
            print(tables['products'].select_dtypes('number').agg([min, max]))
            
            # for col in tables['products'].select_dtypes('number'):
            #     print(f"minimum {col} is {tables['products'][col].min()}\nmaximum {col} is {tables['products'][col].max()}")
            print()
            print()

        if 'sales' in tables:
            print(tables['sales'].select_dtypes('number').agg([min, max]))

    def mean_median_values(self):
        for name, df in tables.items():
            print(f"In {name} table")
            print(f"{df.select_dtypes('number').agg(["mean", "median"])}")
            print()
            print()
            
obj = data_summary()
obj.show_numerical_col()
```

## 4. Missing Values
Check missing values in all three tables.

Note which columns have missing data.

```python
class missing_values():
    def missing(self):
        for name, df in tables.items():
            print(f"In {name} table\nmissing value count is\n{df.isnull().sum()}")
            print()
obj = missing_values()
obj.missing()
```

## 5. Duplicates
Check duplicate rows in each table.

Note the result.

```python
class duplicated_values():
    def duplicated_values(self):
        for name, df in tables.items():
            print(f"In {name} table\nmissing value count is: {df.duplicated().sum()}")
            print()

obj = duplicated_values()
obj.duplicated_values()
```
